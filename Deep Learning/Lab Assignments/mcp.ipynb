```python
def g(x):
    """Activation function: Step function."""
    return 1 if x >= 0 else 0

# Input data and expected outputs for the AND function
inputs = [[0, 0], [0, 1], [1, 0], [1, 1]]
outputs = [0, 0, 0, 1]
weights = [0.5, 0, 1]  # Initial weights
n = 0.25  # Learning rate
iterations = 0

# Training loop
while True:
    iterations += 1
    old_weights = weights.copy()
    for (x, y), op in zip(inputs, outputs):
        val = g(weights[0] + weights[1] * x + weights[2] * y)
        if val != op:  # Update weights if prediction is incorrect
            weights[0] += n * (op - val)  # Update bias weight
            weights[1] += n * (op - val) * x  # Update weight for x
            weights[2] += n * (op - val) * y  # Update weight for y
        print(weights, '\t', val)  # Print weights and predicted value
    print(old_weights, '\t', weights, '\n')  # Print old and new weights for comparison

    if old_weights == weights:  # Stop if weights are stable
        break

print("Final Weights:", weights)
print("Number of Iterations:", iterations)
```
